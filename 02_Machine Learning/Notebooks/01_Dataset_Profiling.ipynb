# Dataset Profiling

## Objective

The objective of this notebook is to analyze the collected training datasets before creating the master dataset. This includes examining the dataset structure, checking missing values, duplicate records, data types, statistical summaries, and overall data quality. The analysis will help determine which datasets are suitable for merging and further preprocessing for machine learning.

Dataset Profiling

Dataset profiling is the process of analyzing a dataset to understand its quality, completeness, structure, and statistical properties before preprocessing and model training.

After profiling, we identified the number of rows, columns, missing values, duplicate records, and data types of each dataset. This information will help us create the master dataset in the next phase.
