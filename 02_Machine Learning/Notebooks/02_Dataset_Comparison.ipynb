# Dataset Comparison

## Objective

The objective of this notebook is to compare all the selected training datasets based on their structure, available water quality parameters, target variables, number of records, and data completeness. This comparison helps identify the contribution of each dataset and provides a clear strategy for creating a unified master dataset for machine learning.

## Theory

Different datasets collected from various sources contain different combinations of water quality parameters. Before creating a master dataset, it is essential to compare these datasets to identify common and unique parameters, missing values, target variables, and overall data quality. This comparison ensures that only relevant datasets and parameters are considered while creating the final master dataset, reducing complexity and improving consistency.

```python
import pandas as pd
import numpy as np
import os
```

```python
dataset_folder = "../../01_Datasets/Training Datasets"

dataset_files = sorted(
    [file for file in os.listdir(dataset_folder) if file.endswith(".csv")]
)

print("Training Datasets:\n")

for file in dataset_files:
    print(file)
```

```python
# Dictionary to store all datasets

datasets = {}

for file in dataset_files:
    file_path = os.path.join(dataset_folder, file)
    datasets[file] = pd.read_csv(file_path)

print("All datasets loaded successfully!")
```

```python
print("Datasets Loaded:\n")

for name in datasets:
    print(name)
```

```python
comparison = []
```

```python
for name, df in datasets.items():

    rows = df.shape[0]
    columns = df.shape[1]

    comparison.append({
        "Dataset": name,
        "Rows": rows,
        "Columns": columns
    })
```

```python
comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

for name, df in datasets.items():

    comparison.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Target": "Yes" if "Target" in df.columns else "No",
        "WQI": "Yes" if "WQI" in df.columns else "No"
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

for name, df in datasets.items():

    comparison.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Duplicate Rows": df.duplicated().sum(),
        "Target": "Yes" if "Target" in df.columns else "No",
        "WQI": "Yes" if "WQI" in df.columns else "No"
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

for name, df in datasets.items():

    comparison.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum(),
        "Target": "Yes" if "Target" in df.columns else "No",
        "WQI": "Yes" if "WQI" in df.columns else "No"
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

# Standard parameters selected for our project
standard_parameters = [
    "pH",
    "Temperature",
    "Turbidity",
    "Total Dissolved Solids",
    "Conductivity",
    "Dissolved Oxygen",
    "Hardness",
    "Nitrate",
    "Chloride",
    "BOD",
    "Target",
    "WQI"
]

for name, df in datasets.items():

    available = []

    for parameter in standard_parameters:
        if parameter in df.columns:
            available.append(parameter)

    comparison.append({
        "Dataset": name,
        "Parameters Present": ", ".join(available)
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

for name, df in datasets.items():

    missing = []

    for parameter in standard_parameters:
        if parameter not in df.columns:
            missing.append(parameter)

    comparison.append({
        "Dataset": name,
        "Missing Parameters": ", ".join(missing)
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison = []

for name, df in datasets.items():

    available = []
    missing = []

    for parameter in standard_parameters:

        if parameter in df.columns:
            available.append(parameter)
        else:
            missing.append(parameter)

    comparison.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing Values": df.isnull().sum().sum(),

        "Duplicate Rows": df.duplicated().sum(),

        "Target": "Yes" if "Target" in df.columns else "No",

        "WQI": "Yes" if "WQI" in df.columns else "No",

        "Parameters Present": ", ".join(available),

        "Missing Parameters": ", ".join(missing)

    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
comparison_df.to_csv(
    "../../08_Documentation/Dataset_Comparison_Report.csv",
    index=False
)

print("Dataset Comparison Report Saved Successfully!")
```
