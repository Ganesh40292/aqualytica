# Dataset Selection Strategy

## Objective

The objective of this notebook is to determine the most appropriate strategy for selecting data from the available training datasets before creating the master dataset. Since the collected datasets differ significantly in size, parameter availability, and target variables, this analysis aims to ensure that the final master dataset is balanced, representative, and suitable for machine learning. The decisions made in this notebook will define how much data is taken from each dataset and whether a dataset should be fully included, partially sampled, or used only as a reference.

## Theory

In real-world machine learning projects, datasets collected from multiple sources often vary in size and quality. Directly combining all datasets without analysis may lead to data imbalance, where one large dataset dominates the training process and reduces the contribution of smaller datasets. Therefore, before creating the master dataset, it is important to evaluate each dataset based on its size, available parameters, target variables, and overall usefulness. A well-planned dataset selection strategy helps improve model performance, ensures fair representation of different data sources, and creates a more reliable dataset for training and evaluation.

```python
import pandas as pd
import numpy as np
import os
```

```python
dataset_folder = "../../01_Datasets/Training Datasets"

dataset_files = sorted(
    [file for file in os.listdir(dataset_folder) if file.endswith(".csv")]
)

datasets = {}

for file in dataset_files:
    datasets[file] = pd.read_csv(os.path.join(dataset_folder, file))

print("All datasets loaded successfully!")
```

```python
summary = []

for name, df in datasets.items():

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing Values": df.isnull().sum().sum(),

        "Duplicate Rows": df.duplicated().sum(),

        "Target Available": "Yes" if "Target" in df.columns else "No",

        "WQI Available": "Yes" if "WQI" in df.columns else "No"

    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
summary = []

for name, df in datasets.items():

    total_cells = df.shape[0] * df.shape[1]

    missing = df.isnull().sum().sum()

    missing_percentage = (missing / total_cells) * 100

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing %": round(missing_percentage,2)

    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
standard_parameters = [

    "pH",

    "Temperature",

    "Turbidity",

    "Total Dissolved Solids",

    "Conductivity",

    "Dissolved Oxygen",

    "Hardness",

    "Nitrate",

    "Chloride",

    "BOD"

]

summary = []

for name, df in datasets.items():

    available = 0

    for parameter in standard_parameters:

        if parameter in df.columns:

            if df[parameter].notna().sum() > 0:

                available += 1

    summary.append({

        "Dataset": name,

        "Parameters with Values": available,

        "Total Parameters": len(standard_parameters)

    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
summary = []

for name, df in datasets.items():

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0]

    })

summary_df = pd.DataFrame(summary)

summary_df = summary_df.sort_values(by="Rows", ascending=False)

summary_df
```
