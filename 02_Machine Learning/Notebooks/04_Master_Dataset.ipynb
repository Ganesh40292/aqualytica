# Master Dataset Creation

## Objective

The objective of this notebook is to create a standardized master dataset by integrating all selected training datasets into a common structure. Each dataset contributes different water quality parameters collected from different sources. During this stage, all datasets are aligned to a common schema, merged sequentially, validated after every merge, and intermediate versions are saved. The resulting master dataset will serve as the primary dataset for preprocessing and machine learning model development.

## Theory

A master dataset is created by combining multiple datasets into one standardized dataset. Since different datasets contain different water quality parameters and target variables, all datasets are first aligned to a common schema before merging. This approach ensures consistency in column names, simplifies future preprocessing, and provides a reliable dataset for machine learning. Missing values are intentionally preserved during this stage and will be handled during preprocessing.

```python
import pandas as pd
import numpy as np
import os
```

```python
dataset_folder = "../../01_Datasets/Training Datasets"

dataset_files = sorted(
    [f for f in os.listdir(dataset_folder) if f.endswith(".csv")]
)

print("Training Datasets Found:\n")

for file in dataset_files:
    print(file)
```

```python
datasets = {}

for file in dataset_files:

    path = os.path.join(dataset_folder, file)

    datasets[file] = pd.read_csv(path)

print("All datasets loaded successfully!")
```

```python
MASTER_COLUMNS = [

    "pH",

    "Temperature",

    "Turbidity",

    "Total Dissolved Solids",

    "Conductivity",

    "Dissolved Oxygen",

    "Hardness",

    "Nitrate",

    "Chloride",

    "BOD",

    "Target",

    "WQI"

]
```

```python
output_folder = "../../01_Datasets/Master Dataset"

os.makedirs(output_folder, exist_ok=True)

print("Output folder ready.")
```

```python
master_dataset = pd.DataFrame()

version = 1

for file in dataset_files:

    print("="*60)
    print(f"Merging {file}")

    df = datasets[file].copy()

    # Add missing columns
    for column in MASTER_COLUMNS:

        if column not in df.columns:

            df[column] = np.nan

    # Arrange columns in standard order
    df = df[MASTER_COLUMNS]

    # Merge
    master_dataset = pd.concat(
        [master_dataset, df],
        ignore_index=True
    )

    print("Rows :", master_dataset.shape[0])
    print("Columns :", master_dataset.shape[1])

    # Save Version
    save_path = os.path.join(
        output_folder,
        f"Master_Dataset_v{version}.csv"
    )

    master_dataset.to_csv(save_path, index=False)

    print(f"Saved : Master_Dataset_v{version}.csv")

    version += 1
```
