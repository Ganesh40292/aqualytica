# Data Preprocessing

## Objective

The objective of this notebook is to prepare the master dataset for machine learning by identifying and handling missing values, duplicate records, inconsistent data, and other data quality issues. The preprocessing steps performed in this notebook will improve the quality of the dataset and ensure that it is suitable for exploratory data analysis, feature engineering, and machine learning model training.

## Theory

Data preprocessing is one of the most important stages in a machine learning pipeline. Real-world datasets often contain missing values, duplicate records, inconsistent formats, and noisy data. These issues can negatively affect the performance of machine learning models. Therefore, preprocessing is performed to improve data quality before model training while preserving meaningful information.

```python
import pandas as pd
import numpy as np
import os
```

```python
master_path = "../../01_Datasets/Master Dataset/Master_Dataset_v6.csv"

df = pd.read_csv(
    master_path,
    low_memory=False
)

print("Master Dataset Loaded Successfully!")
```

```python
# Count missing values in each column

missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage.round(2)

missing_percentage
```

```python
missing_report = pd.DataFrame({

    "Missing Values": df.isnull().sum(),

    "Missing Percentage": ((df.isnull().sum() / len(df)) * 100).round(2)

})

missing_report
```

```python
# Check relationship between Target and WQI

df.groupby("Target")["WQI"].describe()
```

```python
df = df.drop(columns=["Hardness", "BOD"])

print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
duplicate_percentage = (df.duplicated().sum() / len(df)) * 100

print("Duplicate Rows :", df.duplicated().sum())

print("Duplicate Percentage :", round(duplicate_percentage,2),"%")
```

```python
df = df.drop_duplicates()

print(df.shape)
```

```python
print("Remaining Duplicates :", df.duplicated().sum())
```

```python
numerical_columns = [

    "pH",

    "Temperature",

    "Turbidity",

    "Total Dissolved Solids",

    "Conductivity",

    "Dissolved Oxygen",

    "Nitrate",

    "Chloride",

    "WQI"

]

print(df[numerical_columns].skew())
```

```python
df[numerical_columns].describe()
```

```python
missing_report = pd.DataFrame({

    "Missing Values": df.isnull().sum(),

    "Missing Percentage":
        (df.isnull().sum()/len(df)*100).round(2)

})

missing_report
```

```python
median_columns = [

    "pH",

    "Temperature",

    "Turbidity",

    "Conductivity",

    "Nitrate",

    "Chloride"

]

for column in median_columns:

    df[column] = df[column].fillna(
        df[column].median()
    )

print("Median Imputation Completed")
```

```python
df["Total Dissolved Solids"] = df["Total Dissolved Solids"].fillna(

    df["Total Dissolved Solids"].mean()

)

print("Mean Imputation Completed Successfully!")
```

```python
missing_report = pd.DataFrame({

    "Missing Values": df.isnull().sum(),

    "Missing Percentage":
        (df.isnull().sum() / len(df) * 100).round(2)

})

missing_report
```

```python
print("Dataset Shape:")

print(df.shape)
```

```python
print(df.dtypes)
```

```python
output_folder = "../../01_Datasets/Preprocessed Dataset"

os.makedirs(output_folder, exist_ok=True)

save_path = os.path.join(
    output_folder,
    "Master_Dataset_Preprocessed.csv"
)

df.to_csv(
    save_path,
    index=False
)

print("Preprocessed Dataset Saved Successfully!")
print("Location:", save_path)
```

# Conclusion

In this notebook, the master dataset was successfully preprocessed to improve its quality for machine learning. Columns with extremely high missing values (Hardness and BOD) were removed, duplicate records were eliminated, and missing values in important numerical features were handled using appropriate statistical imputation methods. Median imputation was applied to skewed features, while mean imputation was used for the approximately symmetric Total Dissolved Solids feature. The preprocessed dataset was validated and saved successfully, making it ready for exploratory data analysis and feature engineering.
