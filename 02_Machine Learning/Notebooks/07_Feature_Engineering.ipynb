# Feature Engineering

## Objective

The objective of this notebook is to prepare the preprocessed water quality dataset for machine learning by selecting relevant features, transforming variables when necessary, and creating a final dataset suitable for model training and evaluation.

## Theory

Feature engineering is the process of selecting, transforming, and preparing input variables for machine learning models. Proper feature engineering improves model performance by ensuring that relevant information is retained while unnecessary or inconsistent features are removed.

```python
import pandas as pd
import numpy as np
import os
```

```python
dataset_path = "../../01_Datasets/Preprocessed Dataset/Master_Dataset_Preprocessed.csv"

df = pd.read_csv(
    dataset_path,
    low_memory=False
)

print("Preprocessed Dataset Loaded Successfully!")
```

```python
print("Dataset Shape:")
print(df.shape)

df.info()
```

```python
df.head()
```

```python
print(df["Target"].unique())
```

```python
quality_labels = ["Good", "Moderate", "Poor"]

quality_df = df[
    df["Target"].isin(quality_labels)
].copy()

print("Quality Dataset Shape:")
print(quality_df.shape)
```

```python
quality_df["Target"].value_counts()
```

```python
training_folder = "../../02_Machine Learning/Training"

os.makedirs(training_folder, exist_ok=True)

quality_df.to_csv(
    os.path.join(
        training_folder,
        "Quality_Model_Dataset.csv"
    ),
    index=False
)

print("Quality Dataset Saved Successfully!")
```

```python
potability_labels = ["0", "1"]

potability_df = df[
    df["Target"].isin(potability_labels)
].copy()

print("Potability Dataset Shape:")
print(potability_df.shape)
```

```python
potability_df["Target"].value_counts()
```

```python
potability_df.to_csv(
    os.path.join(
        training_folder,
        "Potability_Model_Dataset.csv"
    ),
    index=False
)

print("Potability Dataset Saved Successfully!")
```

```python
comparison = pd.DataFrame({

    "Dataset": [
        "Quality Model",
        "Potability Model"
    ],

    "Rows": [
        len(quality_df),
        len(potability_df)
    ],

    "Columns": [
        quality_df.shape[1],
        potability_df.shape[1]
    ]

})

comparison
```

```python
table_folder = "../../08_Documentation/Tables"

os.makedirs(table_folder, exist_ok=True)

comparison.to_csv(
    os.path.join(
        table_folder,
        "Model_Dataset_Comparison.csv"
    ),
    index=False
)

print("Comparison Table Saved Successfully!")
```

```python
potability_path = "../../02_Machine Learning/Training/Potability_Model_Dataset.csv"

potability_df = pd.read_csv(
    potability_path,
    low_memory=False
)

print("Potability Dataset Loaded Successfully!")
```

```python
print(potability_df.shape)

potability_df.info()
```

```python
missing_report = pd.DataFrame({

    "Missing Values": potability_df.isnull().sum(),

    "Missing Percentage":
    (potability_df.isnull().sum()/len(potability_df)*100).round(2)

})

missing_report
```

```python
potability_df["Target"] = potability_df["Target"].astype(int)

print(potability_df["Target"].dtype)
```

```python
feature_columns = [

    "pH",
    "Temperature",
    "Turbidity",
    "Total Dissolved Solids",
    "Conductivity",
    "Dissolved Oxygen",
    "Nitrate",
    "Chloride"

]

X = potability_df[feature_columns]

y = potability_df["Target"]

print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```
