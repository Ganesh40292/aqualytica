# Model Deployment

## Objective

The objective of this notebook is to deploy the trained Random Forest model by loading the saved machine learning model and testing predictions on new water quality samples. The deployed model will later be integrated with the Spring Boot backend and React dashboard for real-time water potability prediction.

## Theory

Model deployment is the process of using a trained machine learning model to make predictions on new unseen data. After selecting the best-performing Random Forest classifier, the model is loaded and tested using new sensor values before integrating it into the backend application.

```python
import pandas as pd

import numpy as np

import joblib
```

```python
model_path = "../../02_Machine Learning/Saved Models/Water_Potability_RF_Model.pkl"

rf_model = joblib.load(model_path)

print("Random Forest Model Loaded Successfully!")
```

```python
scaler_path = "../../02_Machine Learning/Saved Models/Feature_Scaler.pkl"

scaler = joblib.load(scaler_path)

print("Scaler Loaded Successfully!")
```

```python
print(type(rf_model))
```

```python
sample = pd.DataFrame({

    "pH":[7.4],

    "Temperature":[24.8],

    "Turbidity":[0.8],

    "Total Dissolved Solids":[280],

    "Conductivity":[430],

    "Nitrate":[5.5],

    "Chloride":[175]

})

sample
```

```python
prediction = rf_model.predict(scaler.transform(sample))

prediction
```

```python
if prediction[0] == 1:

    print("Water is Potable")

else:

    print("Water is Not Potable")
```

```python
sample2 = pd.DataFrame({

    "pH":[8.1],

    "Temperature":[18],

    "Turbidity":[0.2],

    "Total Dissolved Solids":[180],

    "Conductivity":[390],

    "Nitrate":[3],

    "Chloride":[120]

})

prediction2 = rf_model.predict(scaler.transform(sample2))

prediction2
```

```python
if prediction2[0] == 1:

    print("Water is Potable")

else:

    print("Water is Not Potable")
```

```python
samples = pd.DataFrame({

    "pH": [7.2, 7.5, 8.3, 6.8, 7.0],

    "Temperature": [22, 18, 25, 28, 20],

    "Turbidity": [0.2, 0.4, 1.8, 3.5, 0.1],

    "Total Dissolved Solids": [180, 220, 650, 950, 120],

    "Conductivity": [320, 410, 780, 1200, 250],

    "Nitrate": [2, 5, 18, 28, 1],

    "Chloride": [90, 140, 260, 420, 70]

})

samples
```

```python
predictions = rf_model.predict(scaler.transform(samples))

predictions
```

```python
for i, prediction in enumerate(predictions, start=1):

    if prediction == 1:

        print(f"Sample {i}: Water is Potable")

    else:

        print(f"Sample {i}: Water is Not Potable")
```

```python
probabilities = rf_model.predict_proba(scaler.transform(samples))

probabilities
```

```python
for i in range(len(samples)):

    prediction = rf_model.predict(scaler.transform(samples.iloc[[i]]))[0]

    probability = rf_model.predict_proba(scaler.transform(samples.iloc[[i]]))[0]

    print("=" * 50)

    print(f"Sample {i+1}")

    if prediction == 1:
        print("Prediction : Potable")
    else:
        print("Prediction : Not Potable")

    print(f"Confidence : {max(probability)*100:.2f}%")
```

# Conclusion

In this notebook, the trained Random Forest model and its associated preprocessing components were successfully loaded and tested for deployment. The model was used to predict the potability of new water samples, and prediction probabilities were generated to measure the confidence of each prediction. The deployment process confirmed that the saved model performs correctly on unseen data and is ready to be integrated with the Spring Boot backend, React dashboard, and ESP32-based sensing system for real-time water quality monitoring and prediction.
